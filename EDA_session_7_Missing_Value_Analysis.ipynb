**How to create a Null Values in a data**

```python
import numpy as np
import pandas as pd
```

```python
d={'Names':['Ramesh','Suresh',None,'Mahesh'],
   'Age':[None,22,24,25],
   'City':['Hyd','Blr','Pune',None]}
df1=pd.DataFrame(d)
df1
```

```python
df1.to_csv('data1.csv',index=False)
```

```python
d={'Names':['Ramesh','Suresh',np.nan,'Mahesh'],
   'Age':[np.nan,22,24,25],
   'City':['Hyd','Blr','Pune',np.nan]}
df2=pd.DataFrame(d)
df2
```

```python
df2.to_csv('data2.csv',index=False)
```

**KNNImputer**

- KNN: K-nearest neighbours
- where k is a number
- instead of taking avg of all the values data 
   - we will find the neighbours who are close with missing data row
   - by using euclidian  distance calculation will identify the neighbour rows
   - suppose k=2 will identify the two closest rows
   - those closest  rows values will take it will calculate the mean
   - finaly will fill the mean

```python
salary=[50000,np.nan,70000,90000]
age=[25,28,30,40]
df=pd.DataFrame(zip(salary,age),
             columns=['Salary','Age'])
df
```

- sklearn 
  - impute
     - KNNImputer

```python
from sklearn.impute import KNNImputer
imputer=KNNImputer(n_neighbors=2)
pd.DataFrame(imputer.fit_transform(df))
```
