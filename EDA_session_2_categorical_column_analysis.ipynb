```python
# num of applicants from 'Asia'
visa_df # df
visa_df['continent'] # one col
con=visa_df['continent']=='Asia' # one class of that one col
np.sum(con) # count of asia  
visa_df[con] #  extraction of one class of data 
# need to see the data inside the asia total names
```

```python
cont_class=visa_df['continent'].unique()
freq=[]
for i in cont_class:
    con=visa_df['continent']==i
    freq.append(np.sum(con))
pd.DataFrame(zip(cont_class,freq),columns=['Class','Frequency'])
```
